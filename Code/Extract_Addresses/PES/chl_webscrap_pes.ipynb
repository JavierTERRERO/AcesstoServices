```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException
import pandas as pd 
import os
from lxml import html 
import time
```

#### Connect to webdriver and extract data

```python
# edge driver path
edge_driver_path = 'C:/Users/Salazar-lozada_M/Downloads/edgedriver_win64/msedgedriver.exe'

# Create a new Edge browser instance
edge_options = webdriver.EdgeOptions()
driver = webdriver.Edge()

# wait 5 seconds for the elements to appear 
driver.implicitly_wait(5)

# Specifying url
url = "https://www.bne.cl/listado-omil" 
driver.get(url)

# Creating lists in order to store data
city = []
state = []
street_address = []

while True: 
    try:
        # Perform web scraping here on the current page
        city_elements = driver.find_elements(By.XPATH, '//form[@id="listadoOmilForm"]/div[3]/div/table/tbody/tr/td/table/tbody/tr[1]/td')
        for cities in city_elements:
            text = cities.text
            print(text)
    
            city.append(text)
    
        state_elements = driver.find_elements(By.XPATH, '//form[@id="listadoOmilForm"]/div[3]/div/table/tbody/tr/td/table/tbody/tr[2]/td')
        for states in state_elements:
            text = states.text
            print(text)
    
            state.append(text)
    
        street_address_elements = driver.find_elements(By.XPATH, '//form[@id="listadoOmilForm"]/div[3]/div/table/tbody/tr/td/table/tbody/tr[3]/td')
        for street_addresses in street_address_elements:
            text = street_addresses.text
            print(text)
    
            street_address.append(text)
        
        # Find and click the element
        next_page = driver.find_element(By.CSS_SELECTOR, '[data-pagina="next"]')
        next_page.click()
        
    except NoSuchElementException:
        break

# Creating dataframes from each list         
city = pd.DataFrame(city)
state = pd.DataFrame(state)
street_address = pd.DataFrame(street_address)

# joining the data together into one dataframe
df = pd.concat([street_address, city, state], axis=1)

df.columns = ['address', 'city', 'state']

df.to_excel('C:/Users/Salazar-lozada_M/downloads/chl_pes.xlsx', index=False)
```
