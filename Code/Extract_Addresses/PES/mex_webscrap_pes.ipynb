```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException
import pandas as pd 
import os
from lxml import html 
import time
```

#### Connect webdriver to the url and extract the data 

```python
# edge driver path
edge_driver_path = 'C:/Users/Salazar-lozada_M/Downloads/edgedriver_win64/msedgedriver.exe'

# Create a new Edge browser instance
edge_options = webdriver.EdgeOptions()
driver = webdriver.Edge()

# wait 5 seconds for the elements to appear 
driver.implicitly_wait(5)

# Specifying url
url = "https://www.empleo.gob.mx/necesitas-ayuda" 
driver.get(url)

# Creating lists in order to store data
state = []
municipality = []
full_address = []

# Finding the website element for the regions dropdown menu and creating a select element
region = driver.find_element(By.XPATH, '//div[@class="col-6 text-left mb-4"]/select')
select = Select(region)

# looping through the regions in the dropdown menu
for option in select.options:
    # Get the text of the option
    region = option.text
    
    # Click the option to select it
    option.click()
    
    time.sleep(2)
    
    # finding the website element for the municipality dropdown menu and creating a select element
    municipal = driver.find_element(By.XPATH, '//div[@class="col-md-6 text-left mb-4"]/select')
    select_1 = Select(municipal)
    
    # looping through each municipality in the dropdown menu 
    for option in select_1.options: 
        municipal = option.text
    
        # Click the option to select it
        option.click() 
        
        time.sleep(2)
        
        # creating a try and error method to find the address details, if they do not appear in the website we're telling the
        # code to not break down and instead to print "Address not found on this page" and to continue
        try:
            address_element = driver.find_element(By.XPATH, '//section[@class="container mt-3"]/div/div/div[2]/div[2]/p')
            address = address_element.text
        
            print("Address:", address)
        except NoSuchElementException:
            print("Address not found on this page")
        
        # appending the webscraped data to the lists we created at the beginning 
        state.append(region)
        municipality.append(municipal)
        full_address.append(address)

# Creating dataframes from each list         
state = pd.DataFrame(state)
municipality = pd.DataFrame(municipality)
address = pd.DataFrame(full_address)

# joining the data together into one dataframe
df = pd.concat([address, municipality, state], axis=1)

# cleaning data (giving each column a name, dropping first row and dropping duplicates)
df.columns = ['address', 'municipality', 'state']
df = df.drop(1)
df = df.drop_duplicates(subset=['address'])

df.to_excel('C:/Users/Salazar-lozada_M/OneDrive - OECD/Work/indicators/pes-location-raw/MEX/mex_pes.xlsx', index=False)
```
