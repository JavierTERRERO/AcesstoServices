### Colombia Data Cleaning 

Since the data was downloaded the only thing needed to do is the following 

1. Clean each dataset 
    i. the private PES offices are sometimes virtual (removed 111 observations)
        a. There are restricted PES offices that are usually for student, these are not removed
    ii. the public PES offices are sometimes mobile, virtual or information centers (removed 96 observations) 
2. remove columns that are not needed 
3. merging the two datasets: private and public PES centers 
4. check for duplicates and remove them if there are any

```python
import pandas as pd 

private = pd.read_excel(r"V:\SPATIAL_INEQUALITIES\BACKUP\Access to services\spatial_inequalities_phaseII\PES\Data\raw-data\COL\prestadores_privados.xlsx")
public = pd.read_excel(r"V:\SPATIAL_INEQUALITIES\BACKUP\Access to services\spatial_inequalities_phaseII\PES\Data\raw-data\COL\prestadores_publicos.xlsx")

# 1. Cleaning each dataset
# printing the different types of PES offices in the private database
#agency_type = private['Tipo de punto de atención'].unique()
#for value in agency_type: 
#    print(value)

# printing the different types of PES offices in the public database
#agency_type = public['Tipo de punto de atención'].unique()
#for value in agency_type: 
#    print(value)

# Removing the agency types that are virtual from the private data 
private = private[~private['Tipo de punto de atención'].isin(['PUNTO VIRTUAL', 'PUNTO VIRTUAL RESTRINGIDO'])]

# Removing the agency types that are virtual, mobile or an information center from the public data
public = public[~public['Tipo de punto de atención'].isin(['BRIGADA MOVIL', 'VEHICULO', 'PUNTO DE INFORMACIÓN', 'PUNTO VIRTUAL'])]

##########################################################################################################
# 2. Removing columns that are not needed 

# Removing all columns that do not serve any purpose for us 
private = private[['Tipo de punto de atención', 'Departamento', 'Municipio', 'Dirección']]

# Removing all columns that do not serve any purpose for us 
public = public[['Tipo de punto de atención', 'Departamento', 'Municipio', 'Dirección']]

##########################################################################################################
# 3. Mergining the two datasets

# Merge the two DataFrames based on the 'key' column
df = pd.concat([private, public], axis=0, ignore_index=True)

##########################################################################################################
# 4. check for duplicates and export

duplicates = df[df.duplicated(subset=['Dirección', 'Municipio'], keep=False)]
#print(duplicates)

df.to_csv('V:\SPATIAL_INEQUALITIES\BACKUP\Access to services\spatial_inequalities_phaseII\PES\Data\clean-data\col_pes.csv', index=False)
```
