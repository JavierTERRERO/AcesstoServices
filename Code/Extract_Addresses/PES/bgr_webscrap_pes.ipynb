#### Cleaning data 

Multiple changes are being done in this window. 

1. removing non-address information such as the word "street" or "city" being included in the corresponding cells 
2. Adding missing data from the Ямбол region 
3. removing other aspects of the cells such as "Адрес:" being included in the address column when there's no need for it 
4. removing duplicates from the dataset 

This gives us the final dataset bgr_pes.xlsx that is saved in the spatial_inequalities_phaseII folder 

```python
df = pd.read_excel('C:/Users/Salazar-lozada_M/Downloads/bgr_raw_data.xlsx')

# Replace cells starting with "Телефон" and "Адрес" in the "address" and "postal code" columns, respectively, with "N/A"
df['address'] = df['address'].apply(lambda x: 'N/A' if x.startswith('Телефон') else x)

# Copy and paste postal code into street column when street is "N/A"
df['address'] = df.apply(lambda row: row['postal code'] if row['address'] == 'N/A' else row['address'], axis=1)

# Remove postal code if past code was true
df['postal code'] = df['postal code'].apply(lambda x: '' if x.startswith('Адрес') else x)

# Adding missing data from the Ямбол region 
row_1 = {'office': 'ДБТ Ямбол, филиал Стралджа',
           'address': 'ул. "П.Яворов" 55',
           'postal code': '8680',
           'city': 'гр. Стралджа'}

row_2 = {'office': 'ДБТ Елхово',
           'address': 'гр.Елхово ул. Търговска 31',
           'postal code': '8700',
           'city': 'гр. Елхово'}

row_3 = {'office': 'ДБТ Ямбол',
           'address': 'ул. "Васил Карагьозов" 2',
           'postal code': '8600',
           'city': 'гр. Ямбол'}

row_4 = {'office': 'ДБТ Елхово, филиал Болярово',
           'address': 'ул. "9-ти септември" 37',
           'postal code': '',
           'city': 'гр. Болярово'}

row_5 = {'office': 'ДБТ Ямбол, филиал Тунджа - Тенево',
           'address': '',
           'postal code': '',
           'city': 'с. Тенево'}

# Add the new row to the DataFrame
df = df.append([row_1, row_2, row_3, row_4, row_5], ignore_index=True)

# Remove everything before and including "Адрес: " in the "addres" column
df['address'] = df['address'].apply(lambda x: x.split("Адрес:")[1] if "Адрес:" in x else x)

# Remove everything before and including "ул." in the "address" column
df['address'] = df['address'].apply(lambda x: x.split("ул.")[1] if "ул." in x else x)

# Remove everything before and including ": " in the "postal code" column
df['postal code'] = df['postal code'].apply(lambda x: x.split(": ")[1] if ": " in x else x)

# Remove everything before and including "гр. " in the "city" column
df['city'] = df['city'].apply(lambda x: x.split("гр. ")[1] if "гр. " in x else x)

# Remove everything before and including "c. " in the "city" column
df['city'] = df['city'].apply(lambda x: x.split("с. ")[1] if "с. " in x else x)

# removing the quotation marks in the address column
df['address'] = df['address'].str.replace('"', '')

# dropping duplicates (goes from 276 to 188 observations)
df = df.drop_duplicates(subset=['address', 'postal code'])

df.to_excel('C:/Users/Salazar-lozada_M/Downloads/bgr_pes.xlsx', index=False)
```
