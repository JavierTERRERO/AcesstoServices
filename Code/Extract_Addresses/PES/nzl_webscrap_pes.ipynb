## New Zealand PES Code

```python
import pandas as pd 

df = pd.read_excel(r'V:\SPATIAL_INEQUALITIES\BACKUP\Access to services\spatial_inequalities_phaseII\PES\Data\raw-data\NZL\New Zealand.xlsx')

split_row = 131

# Create two dataframes based on the split row
df1 = df.iloc[:split_row]
df1.columns = df1.iloc[0]
df1 = df1.drop(0)

df1.rename(columns={'Full_Address': 'address'}, inplace=True)
df1 = df1[['address', 'Longitude', 'Latitude']]

df2 = df.iloc[split_row:]
df2.columns = df2.iloc[0]
df2 = df2.drop(131)

df2.rename(columns={'Address': 'address'}, inplace=True)
df2 = df2[['address', 'Longitude', 'Latitude']]

df = pd.concat([df1, df2], axis=0, ignore_index=True)

df = df.dropna(subset=['address'])

df.to_csv(r'V:\SPATIAL_INEQUALITIES\BACKUP\Access to services\spatial_inequalities_phaseII\indicators\pes-location-raw\NZL\nzl_pes.csv', index=False)
```
