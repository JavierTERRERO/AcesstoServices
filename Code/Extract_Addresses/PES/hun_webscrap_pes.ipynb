```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException
from selenium.webdriver.common.keys import Keys
import pandas as pd 
import os
from lxml import html 
import time
```

#### Connect to webdriver and extract data

collecting url's into the office_links list 

```python
# edge driver path
edge_driver_path = 'C:/Users/Salazar-lozada_M/Downloads/edgedriver_win64/msedgedriver.exe'

# Create a new Edge browser instance
edge_options = webdriver.EdgeOptions()
driver = webdriver.Edge()

# wait 5 seconds for the elements to appear 
driver.implicitly_wait(5)

# Specifying url
url = "https://illetekes.munka.hu/" 
driver.get(url)

time.sleep(5)
wait = WebDriverWait(driver, 10)

office_link = []

while True: 
    
    # Creating lists in order to store data
    offices = driver.find_elements(By.XPATH, '//div[@class="col-md-12"]/div[2]/div/table/tbody/tr[2]/td/table/tbody/tr')
    
    for office in offices:  

        office = office.find_element(By.XPATH, './td/a') 
        href_value = office.get_attribute("href")
        office_link.append(href_value)
    
    try:
    
        next_page = driver.find_element(By.XPATH, '//div[@class="col-md-12"]/div[2]/div/table/tbody/tr[3]/td/span/input[3]')
        next_page.click()
        
    except NoSuchElementException: 
        break
        
    driver.quit()

    
```

Clicking through each url in the office_links list

```python
driver = webdriver.Edge()

Full_address = []

for office in office_link:
    
    try: 
        driver.get(office)
        time.sleep(2)
        address_element = driver.find_element(By.XPATH, '//div[@class="col-md-12"]/div[2]/table/tbody/tr[4]/td[2]')
        address = address_element.text
        print(address)
        Full_address.append(address)
    except WebDriverException as e:
        print(f"WebDriverException: {e}")
        print(f"Skipping office: {office}")
        Full_address.append(f"{office}")  # You can replace None with any value that indicates an error
        continue
    
df = pd.DataFrame(Full_address)

df.to_csv('C:/Users/Salazar-lozada_M/downloads/hun_pes.csv', index=False)
```

```python
df = pd.DataFrame(Full_address)
new_column_name = 'address'
df.columns = [new_column_name]
print(f"Number of rows in the DataFrame: {len(df)}")

df = df.drop_duplicates(keep='first')
print(f"Number of rows in non-duplicates DataFrame: {len(df)}")

df[['postal_code', 'town', 'street']] = df['address'].str.split(n=2, expand=True)
df['street'] = df['street'].str.rstrip('.')

df.to_csv('C:\Users\Salazar-Lozada_M\Downloads\hun.pes.csv', index=False, encoding='iso-8859-2')
```
