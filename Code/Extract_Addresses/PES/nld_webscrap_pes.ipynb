### Parsing data

```python
# define a function to parse URLs
def parse_url(url):
    return url.split('/')[3:]

# apply the parse_url function to the URLs column of the DataFrame
df['parsed_url'] = df['url'].apply(parse_url)

# print the updated DataFrame
print(df)
```

### Seperating the columns with lists values 

```python
# create separate columns for each element in the list
df[['Column1', 'Column2', 'Column3', 'Column4', 'Column5', 'Column6']] = df['parsed_url'].apply(lambda x: pd.Series(x))

print(df)
```

### Clean up addresses column

```python
# replace '+' with ' ' in each cell of the column
df['Column4'] = df['Column4'].str.replace('+', ' ')

print(df)
```

### Creating new columns for elements of the address 

```python
# renaming address column 
df.rename(columns = {'Column4':'address'}, inplace = True)

# extract street names and numbers
df['street'] = df['address'].apply(lambda x: re.findall(r'^\D*\d+\s*\D+', x)[0].strip())
df['street'] = df['street'].str.split(',', n=1, expand=True)[0]


# extract zipcodes
df['zipcode'] = df['address'].str.extract(r'\b(\d{4}\s?[A-Z]{2})\b')

# extract cities
df['city'] = df['address'].str.extract(r'\b([A-Z][a-z]+\s?[A-Z]?[a-z]*)$')

# print the resulting DataFrame
print(df)
```

### Extracting longitude and latitude values from url

```python
def extract_substring(url):
    start = url.find("@")
    if start >= 0: # check if "@" is found in the input string
        try:
            end = url.index("/", start)
            return url[start:end]
        except ValueError: # catch the "substring not found" error
            return "" # return an empty string if "/" is not found in the input string
    else:
        return "" # return an empty string if "@" is not found in the input string

df['Column5'] = df['url'].apply(extract_substring) # apply the function to the whole column

print(df)
```

### Seperating Longitude and Latitude

```python
df[['latitude', 'longitude']] = df['Column5'].str.extract(r'@([\d\.]+),([\d\.]+)')

print(df)
```
