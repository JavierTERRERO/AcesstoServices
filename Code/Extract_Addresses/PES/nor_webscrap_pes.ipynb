```python
# You must pip install -- upgrade selenium and a bunch of other commands
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException
from selenium.common.exceptions import ElementClickInterceptedException
import pandas as pd 
import os
from lxml import html 
import time
import csv

# Path to Edge webdriver executable file on your system
edge_driver_path = "C:\\Users\\TERRERO_J\\Downloads\\edgedriver_win64"

# Create a new Edge browser instance
driver = webdriver.Edge()

# Get webpage wait 10 seconds for the elements to appear 

url = "https://www.nav.no/sok?f=0&uf=1"
driver.get(url)

time.sleep(10)

# Creating lists in order to store data

full_address = []



# Go at the bottom of the page so all centres are charged

try:
   while True:
      continuelink = driver.find_element(By.XPATH,"/html/body/div[2]/div/main/div/div[1]/div[2]/button")
      continuelink.click()
      
      time.sleep(5)

except Exception as e:
   print("no more results to show")


   # Not all centres have visible addresses from this webpage. Need to click in each individual NAV centre page. These two lines capture all the links and creates list.

centretags = driver.find_elements(By.XPATH,"//div[2]/div/main/div/div[1]/div[2]/a") 
centre_links = []

    # Iterate through each element and get the href attribute (i.e., the link)
for centretag in centretags:
   centre_link = centretag.get_attribute("href")
   centre_links.append(centre_link)

print(centre_links)
```

```python

# As a next step, we want to retrieve the addresses from each specific centre link. 
driver = webdriver.Edge()
# Define a list of words that if present in the link, the address will be skipped. This is to skip centres that do not offer employment services in the list of links. 
words_to_skip = ['hjelpemiddelsentral', 'medlemskap', 'forvaltningstjenester', 'styringsenheten-for-nav-hjelpemidler-og-tilrettelegging']  

for centre_link in centre_links:
    # Check if any of the words to skip are in the centre link
    if any(word in centre_link for word in words_to_skip):
        print(f"Skipping link due to filtered word: {centre_link}")
        continue  # Skip to the next iteration of the loop if the condition is met
    
    driver.get(centre_link)
    time.sleep(2)  # Wait for the page to load
    
    # ... rest of your existing code ...

    
    if location_links:
        for i in range(len(location_links)):
            # Re-fetch the location links to avoid StaleElementReferenceException
            location_links = driver.find_elements(By.XPATH, "//div/div[2]/div/div[1]/div[1]/div/button")
            try:
                location_links[i].click()
                time.sleep(3)  # Wait for the location details to load
            except ElementClickInterceptedException:
                print(f"ElementClickInterceptedException at {centre_link}")
                continue

            try:
                centre_address = driver.find_element(By.XPATH, "//div/section/p[1]").text
                full_address.append([centre_address])
                print(centre_address, centre_link)
            except NoSuchElementException:
                print(f"No address found at {centre_link}")
            
        time.sleep(3)

    else:

        try:
                centre_address = driver.find_element(By.XPATH, "//div/section/p[1]").text
                full_address.append([centre_address])
                print(centre_address, centre_link)

        except NoSuchElementException:
                centre_address = driver.find_element(By.XPATH, "//div/div/div/div[1]/div/article/div[1]/p").text
                full_address.append([centre_address])
                print(centre_address, centre_link)

        except NoSuchElementException:
                print(centre_address, centre_link)
   


# Close the WebDriver
driver.quit()

```

```python
## Note that for around 100 centres there are no "visiting" addresses, only postal boxes. In those cases the output returned is only a mobile number. 

# We suspect that this is the case for only administrative offices, i.e., offices that do not offer in-person services. 

#This line of command filters these offices (for which we only have phone number) and removes them from the database. 

filtered_addresses = [address for address in full_address if not address[0].replace(" ", "").replace(",", "").isnumeric()]
print (filtered_addresses)


# Drop duplicates
unique_addresses = list(set([tuple(address) for address in filtered_addresses]))

# Convert back to list of lists from list of tuples
unique_addresses = [list(address) for address in unique_addresses]

print(unique_addresses, centre_link)

```

```python
# Export to CSV
with open('nor_pes.csv', 'w', newline='') as csvfile:
    csvwriter = csv.writer(csvfile)
    
    # Write the header
    csvwriter.writerow(['Address'])
    
    # Write the data
    for address in unique_addresses:
        csvwriter.writerow(address)  # Assuming address is a list with one element
```
